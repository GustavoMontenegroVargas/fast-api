#### GroupBy

````python
import pandas as pd
import numpy as np

# 1. Creamos el dataset de ejemplo
data = {
    'Vendedor': ['Ana', 'Ana', 'Ana', 'Beto', 'Beto', 'Carlos', 'Carlos'],
    'Monto_Venta': [100, 150, 600, 200, 250, 300, 350]
}

data
````

````python
data.keys()
````

````python
df = pd.DataFrame(data)

# --- DIFERENCIA 1: .mean() (Agregación) ---
# Esto reduce el dataset. Útil para un resumen rápido.
resumen_promedios = df.groupby('Vendedor')['Monto_Venta'].mean()

print("--- RESULTADO DE .mean() ---")
print(resumen_promedios)
print(f"Tamaño del resultado: {len(resumen_promedios)}")
````

````python
# --- DIFERENCIA 2: .transform('mean') (Transformación) ---
# Esto mantiene el tamaño original. Útil para Feature Engineering.
df['Promedio_Vendedor'] = df.groupby('Vendedor')['Monto_Venta'].transform('mean')

print("--- RESULTADO DE .transform() ---")
print(df)
print(f"Tamaño del resultado: {len(df)} filas")
````

````python
df.groupby('Vendedor')['Monto_Venta'] #Es un lazzy object, pandas ya sabe cómo va a agrupar los datos, pero no hace ningún cálculo hasta que le pidas una función final.
````

#### Apply

````python
data_apply = {
    "Ticket_ID": ["T1","T2","T3"],
    "Impacto": ["Alto", "Bajo", "Alto"],
    "Horas": [15,20,5]
}

df_tickets = pd.DataFrame(data_apply)

df_tickets
````

````python
# Definimos una función con nuestra lógica de negocio
def categorizar_urgencia(fila):
    if fila['Impacto'] == 'Alto' and fila['Horas'] > 10:
        return 'Crítico'
    else:
        return 'Normal'

# Aplicamos la función a lo largo de las columnas (axis=1)
df_tickets['Prioridad'] = df_tickets.apply(categorizar_urgencia, axis=1)

# Resultado:
# T1 -> Crítico (Alto + 15)
# T2 -> Normal  (Bajo + 20)
# T3 -> Normal  (Alto + 5)

df_tickets
````

````python
data_vendedor = {
    "ID":["A1", "B2"],
    "Nombre":["Juan", "Ana"],
    "Pais":["MX", "US"]
}

data_ventas_sf = {
    "ID":["A1", "A1", "B2"],
    "Producto":["Einstein", "Tableau", "Slack"],
    "Monto":[1000, 500, 2000]
}

# 0. Crear dataframes
df_vendedor = pd.DataFrame(data_vendedor)
df_ventas_sf = pd.DataFrame(data_ventas_sf)

# 1. Merge | Ventas por vendedor
df_ventas_por_vendedor = df_ventas_sf.merge(df_vendedor, on = "ID", how = "left")

# 2. Transform | Suma de ventas por cada persona
df_ventas_por_vendedor["ventas_total"] = df_ventas_por_vendedor.groupby("ID")["Monto"].transform("sum")
# Aunque esta parte es correcta, es siempre mejor agrupar por ID, debido a que son identificadores únicos

# 3. Filtrado por país
df_ventas_por_vendedor_mx = df_ventas_por_vendedor[df_ventas_por_vendedor["Pais"] == "MX"]
````

Ejercicio de Claridad Total

Imagina este DataFrame `df_clientes`:

```

Nombre  Edad  Compras_Anuales
Juan     25       10
Ana      40       50

```

- Misión : 
  - Usa `map()` para crear una columna `Nombre_Mayus` (solo transforma el nombre).
  - Usa `apply()` con axis=1 para crear una columna Segmento:Si Edad > 30 Y Compras_Anuales > 20 -> 'Premium'De lo contrario -> 'Regular'

````python
data_claridad_total = {
    "Nombre": ["Juan", "Ana"],
    "Edad": [25, 40],
    "Compras_Anuales": [10, 50]
}

df_cl = pd.DataFrame(data_claridad_total)
print(df_cl)
print(df_cl.info(memory_usage="deep"))
print(df_cl.memory_usage(deep=True))
````

````python
df_cl["Nombre_Mayus"] = df_cl["Nombre"].map(lambda x:x.upper())
df_cl
````

````python
def definir_segmento(row):
    if row["Edad"] > 30 and row["Compras_Anuales"] > 20:
        # row["Segmento"] = "Premium" | Esta asignación no es requerida.
        return "Premium"
    
    else:
        # row["Segmento"] = "Regular" | Esta asignación no es requerida
        return "Regular"
    

df_cl["Segmento"] = df_cl.apply(definir_segmento, axis = 1)

df_cl
````
